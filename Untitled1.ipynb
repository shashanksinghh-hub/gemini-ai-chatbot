```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

print("Gemini API configured successfully!")
```

```python
def ai_agent(task):
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=f"""
You are a helpful AI Agent.

Your task is:
{task}

Think about the task and provide a helpful response.
"""
    )

    return response.text


task = input("Enter a task for the AI Agent: ")

result = ai_agent(task)

print("\nAI Agent Response:\n")
print(result)
```

```python
def calculator(expression):
    try:
        result = eval(expression)
        return f"Calculation result: {result}"
    except:
        return "Invalid calculation."


question = input("Enter a calculation: ")

print(calculator(question))
```

```python
from datetime import datetime

def get_current_date():
    return datetime.now().strftime("%d-%m-%Y")

def smart_agent(task):

    task_lower = task.lower()

    # Agent decides which tool to use
    if any(char in task for char in ["+", "-", "*", "/"]) or "calculate" in task_lower:
        expression = task_lower.replace("calculate", "").strip()
        return calculator(expression)

    elif "date" in task_lower or "today" in task_lower:
        return f"Today's date is: {get_current_date()}"

    else:
        response = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=f"You are a helpful AI Agent. Answer this task: {task}"
        )
        return response.text


task = input("Give a task to the Smart AI Agent: ")

print("\nAI Agent Response:\n")
print(smart_agent(task))
```

```python
conversation_memory = []

def memory_agent(task):

    # Save user's task in memory
    conversation_memory.append(f"User: {task}")

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=f"""
You are a helpful AI Agent.

Here is the conversation memory:
{conversation_memory}

Current user task:
{task}

Answer the user's current task while considering the conversation memory.
"""
    )

    answer = response.text

    # Save AI response in memory
    conversation_memory.append(f"Agent: {answer}")

    return answer


task = input("Ask the Memory Agent something: ")

print("\nMemory Agent Response:\n")
print(memory_agent(task))
```

```python
task = input("Ask another question: ")

print("\nMemory Agent Response:\n")
print(memory_agent(task))
```

```python
def complete_agent(task):
    task_lower = task.lower()

    print("Step 1: Understanding the task...")

    # Decide which action to take
    if any(char in task for char in ["+", "-", "*", "/"]) or "calculate" in task_lower:
        print("Step 2: Decision → Use Calculator Tool")

        expression = task_lower.replace("calculate", "").strip()
        result = calculator(expression)

        print("Step 3: Processing tool result...")
        return result

    elif "date" in task_lower or "today" in task_lower:
        print("Step 2: Decision → Use Date Tool")

        result = get_current_date()

        print("Step 3: Processing tool result...")
        return f"Today's date is: {result}"

    else:
        print("Step 2: Decision → Ask Gemini AI")

        response = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=f"""
You are a helpful AI Agent.
Answer the following task clearly:

{task}
"""
        )

        print("Step 3: Processing AI response...")
        return response.text


task = input("Give a task to the Complete AI Agent: ")

print("\n--- FINAL RESPONSE ---\n")
print(complete_agent(task))
```
