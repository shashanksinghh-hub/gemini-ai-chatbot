```python
import os
import google.generativeai as genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
genai.configure(api_key=api_key)

print("Gemini API configured successfully!")
```

```python
from google.colab import files
import pypdf

uploaded = files.upload()

pdf_name = list(uploaded.keys())[0]

reader = pypdf.PdfReader(pdf_name)

text = ""
for page in reader.pages:
    text += page.extract_text() + "\n"

print("PDF uploaded successfully!")
print("Total characters:", len(text))
```

```python
model = genai.GenerativeModel("gemini-1.5-flash")

prompt = f"""
You are an AI assistant. Answer questions based on the following PDF content.

PDF CONTENT:
{text}

Give clear and helpful answers.
"""

print("AI is ready!")
```

```python
question = input("Ask a question about the PDF: ")

response = model.generate_content(
    prompt + "\n\nUSER QUESTION: " + question
)

print("\nANSWER:\n")
print(response.text)
```

```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

print("New Gemini AI is ready!")
```

```python
question = input("Ask a question about the PDF: ")

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=f"""
PDF CONTENT:
{text}

USER QUESTION:
{question}

Answer based only on the PDF content.
"""
)

print("\nANSWER:\n")
print(response.text)
```

```python
# Split PDF text into smaller chunks

chunk_size = 1000
chunks = []

for i in range(0, len(text), chunk_size):
    chunk = text[i:i + chunk_size]
    chunks.append(chunk)

print("Total chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
from google import genai

# Create embeddings for all chunks
embeddings = []

for chunk in chunks:
    result = client.models.embed_content(
        model="gemini-embedding-001",
        contents=chunk
    )
    embeddings.append(result.embeddings[0].values)

print("Embeddings created successfully!")
print("Total embeddings:", len(embeddings))
```

```python
import faiss
import numpy as np

# Convert embeddings into NumPy array
embedding_array = np.array(embeddings).astype("float32")

# Create FAISS index
dimension = embedding_array.shape[1]

index = faiss.IndexFlatL2(dimension)

# Add embeddings to Vector Database
index.add(embedding_array)

print("FAISS Vector Database created successfully!")
print("Total vectors stored:", index.ntotal)
```

```python
question = input("Ask a question about the PDF: ")

# Create embedding for user's question
question_result = client.models.embed_content(
    model="gemini-embedding-001",
    contents=question
)

question_embedding = np.array(
    [question_result.embeddings[0].values]
).astype("float32")

# Search for the 3 most relevant chunks
distances, indices = index.search(question_embedding, k=3)

# Get relevant chunks
relevant_chunks = [chunks[i] for i in indices[0]]

print("\nRelevant PDF content found:\n")

for i, chunk in enumerate(relevant_chunks):
    print(f"\n--- Relevant Chunk {i+1} ---\n")
    print(chunk)
```
