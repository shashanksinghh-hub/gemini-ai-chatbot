```python
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model="gemini-3.6-flash",
    config=types.GenerateContentConfig(
        system_instruction="You are a helpful Python teacher. Always explain concepts in simple language with examples."
    ),
    contents="Explain Python lists."
)

print(response.text)
```

```python
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Give me information about Python in JSON format with fields: name, creator, and use.",
    config=types.GenerateContentConfig(
        response_mime_type="application/json"
    )
)

print(response.text)
```

```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

try:
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents="Explain Python exception handling in simple words."
    )
    print(response.text)

except Exception as e:
    print("An error occurred:")
    print(e)
```

```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

chat = client.chats.create(
    model="gemini-3.6-flash"
)

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit", "bye"]:
        print("AI: Goodbye! 👋")
        break

    try:
        response = chat.send_message(user_input)
        print("AI:", response.text)

    except Exception as e:
        print("Error:", e)
```

```python
import gradio as gr
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

def chatbot(message, history):
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=message
    )
    return response.text

demo = gr.ChatInterface(
    fn=chatbot,
    title="🤖 My Gemini AI Chatbot",
    description="Ask me anything!",
    textbox=gr.Textbox(
        placeholder="Type your message here..."
    )
)

demo.launch(share=True, debug=False)
```
